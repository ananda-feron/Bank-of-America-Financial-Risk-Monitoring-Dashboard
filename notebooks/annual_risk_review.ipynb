# Bank of America: annual risk indicators
Run the pipeline before this notebook. Values are USD; ratios are fractions. Missing regulatory ratios are documented, not filled.

```python
from pathlib import Path
import pandas as pd
root = Path.cwd()
if root.name == "notebooks": root = root.parent
annual = pd.read_csv(root / "data/processed/financial_risk_annual.csv")
annual
```

```python
ratios = annual[["fiscal_year", "loan_to_deposit_ratio", "cash_to_assets_ratio", "roa_calculated", "roe_calculated"]].set_index("fiscal_year")
ratios.mul(100).round(2).rename(columns=lambda name: name + "_percent")
```
